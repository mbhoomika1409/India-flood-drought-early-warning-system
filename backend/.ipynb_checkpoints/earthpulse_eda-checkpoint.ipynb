```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
import json

# NASA POWER API - Free climate data
# Let's fetch rainfall + temperature data for Davanagere, Karnataka

latitude = 14.4644
longitude = 75.9218

url = f"https://power.larc.nasa.gov/api/temporal/daily/point?parameters=PRECTOTCORR,T2M,RH2M,WS2M&community=RE&longitude={longitude}&latitude={latitude}&start=19900101&end=20231231&format=JSON"

print("Fetching data from NASA POWER...")
response = requests.get(url)
data = response.json()
print("Done! Status:", response.status_code)
```

```python
# Extract the data into a DataFrame
params = data['properties']['parameter']

df = pd.DataFrame(params)
df.index = pd.to_datetime(df.index, format='%Y%m%d')
df.columns = ['Rainfall', 'Temperature', 'Humidity', 'WindSpeed']

print("Shape:", df.shape)
print("\nFirst 5 rows:")
df.head()
```

```python
# Plot rainfall over the years
plt.figure(figsize=(15, 5))
plt.plot(df.index, df['Rainfall'], color='steelblue', alpha=0.6)
plt.title('Daily Rainfall in Davanagere (1990-2023)')
plt.xlabel('Year')
plt.ylabel('Rainfall (mm)')
plt.tight_layout()
plt.show()
```

```python
# Plot all 4 parameters together
fig, axes = plt.subplots(4, 1, figsize=(15, 12))

df['Rainfall'].plot(ax=axes[0], color='steelblue', title='Rainfall (mm)')
df['Temperature'].plot(ax=axes[1], color='tomato', title='Temperature (°C)')
df['Humidity'].plot(ax=axes[2], color='green', title='Humidity (%)')
df['WindSpeed'].plot(ax=axes[3], color='purple', title='Wind Speed (m/s)')

plt.tight_layout()
plt.show()
```

```python
# Find extreme rainfall months (potential flood periods)
df['Month'] = df.index.month
df['Year'] = df.index.year

# Monthly total rainfall
monthly = df.groupby(['Year', 'Month'])['Rainfall'].sum().reset_index()
monthly.columns = ['Year', 'Month', 'Total_Rainfall']

# Find months where rainfall > 300mm (flood risk threshold)
flood_months = monthly[monthly['Total_Rainfall'] > 300]

print(f"Total flood-risk months found: {len(flood_months)}")
print("\nTop 10 highest rainfall months:")
print(flood_months.sort_values('Total_Rainfall', ascending=False).head(10))
```

```python
# Find drought risk months (very low rainfall in monsoon season)
# Monsoon = June to September (months 6,7,8,9)
monsoon = monthly[monthly['Month'].isin([6,7,8,9])]

# Yearly total monsoon rainfall
yearly_monsoon = monsoon.groupby('Year')['Total_Rainfall'].sum().reset_index()
yearly_monsoon.columns = ['Year', 'Monsoon_Rainfall']

# Average monsoon rainfall
avg = yearly_monsoon['Monsoon_Rainfall'].mean()
print(f"Average monsoon rainfall: {avg:.2f} mm")

# Drought = less than 75% of average
drought_years = yearly_monsoon[yearly_monsoon['Monsoon_Rainfall'] < avg * 0.75]
print(f"\nDrought years found: {len(drought_years)}")
print(drought_years.sort_values('Monsoon_Rainfall'))
```
